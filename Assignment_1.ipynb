```python
#Create list
thislist = ["apple", "banana", "watermelon"]
print("This is your list",thislist)
#access element from list
print("The second element of your list is",thislist[1])
print(thislist[0:2])
#change list items
thislist[1] = "kiwi"
print("Here banana is replaced with kiwi",thislist)
#add list items
thislist.append("orange")
print("Orange is added to the end of the list",thislist)
#remove items from list
thislist.remove("orange")
print("This removes an element",thislist)
#sort lists
thislist = ["apple", "banana", "watermelon", "pineapple", "mango"]
thislist.sort()
print("This sorts the list in an alphabetical order",thislist)
#copying a list
mylist = thislist.copy()
print("You can copy a list with this",mylist)
#joining lists
yourlist = mylist + thislist
print("This joins multiple lists",yourlist)
```

```python
#Create tuple
print("Creating tuple")
thistuple = ("apple", "banana", "cherry")
print(thistuple)
#Access tuple items
print("Accessing tuple items")
thistuple = ("apple", "banana", "cherry")
print(thistuple[1])
#Change tuple values
print("Changing tuple values")
x = ("apple", "banana", "cherry")
y = list(x)
y[1] = "kiwi"
x = tuple(y)
print(x)
#unpacking a tuple
print("Unpacking tuples")
fruits = ("apple", "banana", "cherry")
(green, yellow, red) = fruits
print(green)
print(yellow)
print(red)
#looping through a tuple
print("Looping through a tuple")
thistuple = ("apple", "banana", "cherry")
for x in thistuple:
  print(x)
#joining tuples
print("Joining tuples")
tuple1 = ("a", "b" , "c")
tuple2 = (1, 2, 3)

tuple3 = tuple1 + tuple2
print(tuple3)
```

```python
courses = ("Python", "Data science", "Cloud computing", "Cyber security", "AI")
#Show second course selected
print(courses[1])

#Show last two courses using tuple slicing
coursesl2=courses[3:5]
print("Last two courses are", coursesl2)

#Checking if python is in the tuple
if "Python" in courses:
    print("Python is present in the tuple")
else:
    print("Python is not present in the tuple")
    
#Finding the position of cloud computing
print("The index is",courses.index("Cloud computing"))

#Total number of courses selected
count = len(courses)
print("Total number of courses is", count)

#Adding courses
new_courses = courses + ("Machine learning", "Web development")
print("The new list of courses is", new_courses)

#Create a new tuple without Cyber security
updated_courses = courses[:3] + courses[4:]
print("Courses without Cyber security:", updated_courses)

#Add the new courses
updated_courses = updated_courses + ("Machine learning", "Web development")
print("Updated courses:", updated_courses)

#Arrange courses in alphabetical order
sorted_courses = tuple(sorted(updated_courses))
print("Sorted courses:", sorted_courses)

#Create a backup copy
backup_courses = sorted_courses
print("Backup courses:", backup_courses)

#Friend's selected courses
friend_courses = ("AI", "Block chain", "Python")

#Combining the student's courses and friend's courses
combined_courses = sorted_courses + friend_courses
print("Combined courses:", combined_courses)

#Display the final tuples
print("Final course tuple:", updated_courses)
print("Sorted tuple:", sorted_courses)
print("Backup tuple:", backup_courses)
print("Combined tuple:", combined_courses)
```

```python
#Create dictionary
print("Creating a dictionary")
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(thisdict)

#Accessing Items
print("Accessing items in the dictionary")
x = thisdict["model"]
print(x)

#Changing values in dictionary
print("Changing values in dictionary")
thisdict["year"] = 2018
print(thisdict)

#Adding Items in dictionary
print("Adding items in dictionary")
thisdict["color"] = "red"
print(thisdict)

#Removing Items from dictionary
print("Removing items from dictionary")
thisdict.pop("model")
print(thisdict)

#Looping through a dictionary
print("Looping through a dictionary")
for x in thisdict:
  print(x)

#Copying a dictionary
print("Copying a dictionary")
mydict = thisdict.copy()
print(mydict)

#Creating a nested dictionary
print("Creating a nested dictionary")
myfamily = {"child1" : {"name" : "Emil","year" : 2004},"child2" : {"name" : "Tobias","year" : 2007},"child3" : { "name" : "Linus","year" : 2011}}
print(myfamily)
```
